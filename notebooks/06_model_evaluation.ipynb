## Evaluation philosophy

Metrics must match the target definition.

Because the final target is a future exploit risk target, the evaluation should focus on:
- ranking quality
- recall of high-risk vulnerabilities
- class separation where classification is used
- calibration where appropriate
- robustness to missing values
- feature importance review

This notebook prepares the reusable evaluation structure and will be updated once the final target dataset and model outputs are available.
